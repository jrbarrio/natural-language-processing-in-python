## Introduction to audio data in Python

### Importing an audio file with Python

```python
current_dir = Path.cwd()
print(f"Current working directory: {current_dir}")

# Create audio file wave object
audio_path = current_dir / 'good-morning.wav'
if not audio_path.exists():
    audio_path = current_dir / '04-spoken-language-processing-in-python/1-introduction-to-spoken-language-processing-with-python/good-morning.wav'
good_morning = wave.open(str(audio_path), 'r')

# Read all frames from wave object 
signal_gm = good_morning.readframes(-1)

# View first 10
print(signal_gm[:10])
```
